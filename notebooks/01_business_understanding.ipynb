# 01 — Business Understanding  
## Mandato de Capital de Trabajo (Distribuidores)

---

## 1. Contexto del negocio

Las empresas de distribución operan con márgenes brutos limitados y alta intensidad de capital.  
El principal determinante del desempeño económico **no es el crecimiento de ventas**, sino **cómo se asigna el capital de trabajo**, particularmente en:

- Inventarios (principal consumidor de capital)
- Cuentas por cobrar (financiamiento implícito a clientes)
- Compras y términos con proveedores (rigidez del capital)

En este contexto, pequeñas decisiones operativas mal informadas a nivel SKU pueden destruir ROIC de forma silenciosa.

---

## 2. Problema de negocio

El capital de trabajo está **mal asignado** debido a decisiones fragmentadas y con información superficial:

- Inventarios definidos a nivel categoría o “intuición”
- Falta de visibilidad a nivel SKU / cliente
- Compras optimizadas por precio unitario, no por rotación
- Crédito otorgado sin considerar consumo real de capital

Esto genera:
- Capital atrapado
- ROIC subóptimo
- Falta de claridad para ejecutar decisiones duras

---

## 3. Objetivo del mandato

**Reasignar capital de trabajo para mejorar el desempeño económico del negocio en ciclos cortos (semanas, no años).**

El mandato **no busca crecimiento**, sino:
- Liberar capital donde no genera valor
- Reasignarlo a SKUs, clientes y flujos que sí lo justifican
- Mejorar el ROIC económico del capital operativo

---

## 4. Preguntas clave de negocio

Este análisis debe permitir responder, con evidencia:

1. ¿En qué SKUs está atrapado el capital?
2. ¿Qué inventario no está justificado por demanda real?
3. ¿Qué clientes consumen capital sin retorno suficiente?
4. ¿Qué compras y términos están rigidizando el capital?
5. ¿Qué decisiones concretas deben ejecutarse esta semana?

Si una pregunta no conduce a una decisión ejecutable, no es prioritaria.

---

## 5. Alcance del análisis (MVP)

Incluye:
- Inventario a nivel SKU
- Ventas transaccionales (SKU × cliente)
- Compras y proveedores
- Estimación razonable de CxC y CxP
- ROIC económico del capital operativo

Excluye deliberadamente:
- Crecimiento de ventas
- Marketing
- RRHH
- Optimización de procesos no relacionados con capital

---

## 6. Definición de éxito

El mandato se considera exitoso si:

- El capital de trabajo neto disminuye **o**
- El ROIC económico del capital operativo mejora de forma consistente

La métrica clave no es contable, es **económica y direccional**.

---

## 7. Supuestos explícitos

- Se prioriza precisión direccional sobre exactitud contable
- Inventarios son el principal driver del capital invertido
- El análisis se ejecuta a nivel SKU, no agregado
- Las decisiones deben poder ejecutarse sin renegociación posterior

---

## 8. Resultado esperado de esta fase

Al finalizar esta fase, se debe tener claridad absoluta sobre:

- Qué capital está mal asignado
- Por qué
- Qué decisiones deben tomarse inmediatamente

Todo lo que no contribuya a esto se considera ruido.
